```python
#1.Min-Max Search Algorithm for Chess:

def evaluate(board):
    # Simple evaluation:
    # White pieces = positive
    # Black pieces = negative
    values = {
        'P': 1,
        'N': 3,
        'B': 3,
        'R': 5,
        'Q': 9,
        'K': 100
    }

    score = 0

    for row in board:
        for piece in row:
            if piece in values:
                score += values[piece]
            elif piece.lower() in values:
                score -= values[piece.lower()]

    return score


def generate_moves(board, player):
    moves = []

    for i in range(2):
        for j in range(2):
            if board[i][j] == '.':
                new_board = [row[:] for row in board]
                new_board[i][j] = player
                moves.append(new_board)

    return moves


def minimax(board, depth, maximizing_player):

    if depth == 0:
        return evaluate(board)

    player = 'P' if maximizing_player else 'p'
    moves = generate_moves(board, player)

    if not moves:
        return evaluate(board)

    if maximizing_player:
        best = float('-inf')

        for move in moves:
            value = minimax(move, depth - 1, False)
            best = max(best, value)

        return best

    else:
        best = float('inf')

        for move in moves:
            value = minimax(move, depth - 1, True)
            best = min(best, value)

        return best


#Ex-chess board
board = [
    ['R', 'P'],
    ['p', '.']
]

print("Board:")
for row in board:
    print(row)
result = minimax(board, 3, True)
print("Best evaluation using Min-Max:", result)
```

```python
#2.Alpha-Beta Pruning Algorithm for Chess:

def evaluate(board):
    values = {
        'P': 1,
        'N': 3,
        'B': 3,
        'R': 5,
        'Q': 9,
        'K': 100
    }

    score = 0

    for row in board:
        for piece in row:
            if piece in values:
                score += values[piece]
            elif piece.lower() in values:
                score -= values[piece.lower()]
            return score

def generate_moves(board, player):
    moves = []

    for i in range(2):
        for j in range(2):
            if board[i][j] == '.':
                new_board = [row[:] for row in board]
                new_board[i][j] = player
                moves.append(new_board)

    return moves
def alpha_beta(board, depth, alpha, beta, maximizing_player):

    if depth == 0:
        return evaluate(board)

    player = 'P' if maximizing_player else 'p'
    moves = generate_moves(board, player)

    if not moves:
        return evaluate(board)

    if maximizing_player:

        best = float('-inf')

        for move in moves:

            value = alpha_beta(
                move,
                depth - 1,
                alpha,
                beta,
                False
            )

            best = max(best, value)

            alpha = max(alpha, best)

            # Beta cutoff
            if beta <= alpha:
                break

        return best

    else:

        best = float('inf')

        for move in moves:

            value = alpha_beta(
                move,
                depth - 1,
                alpha,
                beta,
                True
            )

            best = min(best, value)

            beta = min(beta, best)

            # Alpha cutoff
            if beta <= alpha:
                break
        return best


# Ex-chess board
board = [
    ['R', 'P'],
    ['p', '.']
]

print("Board:")
for row in board:
    print(row)
result = alpha_beta(
    board,
    3,
    float('-inf'),
    float('inf'),
    True
)
print("Best evaluation using Alpha-Beta:", result)
```

```python
#3.Min-Max Search Algorithm for Tic-Tac-Toe:

import math


def print_board(board):
    print()
    for i in range(3):
        print(" | ".join(board[i]))
        if i < 2:
            print("--+---+--")
    print()


def check_winner(board):

    # Rows
    for row in board:
        if row[0] == row[1] == row[2] and row[0] != ' ':
            return row[0]

    # Columns
    for col in range(3):
        if board[0][col] == board[1][col] == board[2][col]:
            if board[0][col] != ' ':
                return board[0][col]

    # Diagonals
    if board[0][0] == board[1][1] == board[2][2]:
        if board[0][0] != ' ':
            return board[0][0]

    if board[0][2] == board[1][1] == board[2][0]:
        if board[0][2] != ' ':
            return board[0][2]

    return None


def is_moves_left(board):
    for row in board:
        if ' ' in row:
            return True

    return False


def minimax(board, depth, is_maximizing):

    winner = check_winner(board)

    # Computer wins
    if winner == 'O':
        return 10 - depth

    # Human wins
    if winner == 'X':
        return depth - 10

    # Draw
    if not is_moves_left(board):
        return 0

    if is_maximizing:

        best_score = -math.inf

        for i in range(3):
            for j in range(3):

                if board[i][j] == ' ':

                    board[i][j] = 'O'

                    score = minimax(
                        board,
                        depth + 1,
                        False
                    )

                    board[i][j] = ' '

                    best_score = max(
                        best_score,
                        score
                    )

        return best_score

    else:

        best_score = math.inf

        for i in range(3):
            for j in range(3):

                if board[i][j] == ' ':

                    board[i][j] = 'X'

                    score = minimax(
                        board,
                        depth + 1,
                        True
                    )

                    board[i][j] = ' '

                    best_score = min(
                        best_score,
                        score
                    )

        return best_score


def find_best_move(board):

    best_score = -math.inf
    best_move = None

    for i in range(3):
        for j in range(3):

            if board[i][j] == ' ':

                board[i][j] = 'O'

                score = minimax(
                    board,
                    0,
                    False
                )

                board[i][j] = ' '

                if score > best_score:
                    best_score = score
                    best_move = (i, j)

    return best_move


# Main program

board = [
    [' ', ' ', ' '],
    [' ', ' ', ' '],
    [' ', ' ', ' ']
]

print("TIC-TAC-TOE")
print("You = X")
print("Computer = O")

while True:

    print_board(board)

    # Player move
    row = int(input("Enter row (1-3): ")) - 1
    col = int(input("Enter column (1-3): ")) - 1

    if board[row][col] != ' ':
        print("Position already occupied!")
        continue

    board[row][col] = 'X'

    winner = check_winner(board)

    if winner == 'X':
        print_board(board)
        print("You win!")
        break

    if not is_moves_left(board):
        print_board(board)
        print("It's a draw!")
        break

    # Computer move
    move = find_best_move(board)

    if move:
        board[move[0]][move[1]] = 'O'

    winner = check_winner(board)

    if winner == 'O':
        print_board(board)
        print("Computer wins!")
        break

    if not is_moves_left(board):
        print_board(board)
        print("It's a draw!")
        break
```
